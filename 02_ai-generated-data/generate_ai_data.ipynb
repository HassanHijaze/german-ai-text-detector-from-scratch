# 02 · AI data

For every human sample from step 01, one model writes two texts:

- **AI-assisted**: the human text rewritten and polished
- **AI-generated**: a new text on the same topic

The code is in `ai_generation/`, the prompts in `ai_generation/prompts.py`. API keys go in the `.env` file in the repo root.

```python
from ai_generation import config as cfg, run, find_missing, show_example

for name, spec in cfg.MODELS.items():
    print(f"{name:18} {spec['company']:9} IDs {spec['seed_ids'][0]:>5}-{spec['seed_ids'][1]:<5}  strategy: {spec['strategy']}")
```

## Check what's missing

Failed requests (rate limits, empty answers, no credits left) leave gaps. Running the same model again fills them.

```python
for model, kinds in find_missing().items():
    print(model, {kind: len(ids) for kind, ids in kinds.items()})
```
