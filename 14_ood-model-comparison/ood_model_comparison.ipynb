### OOD Model Comparison

This evaluates all five transformer models on the external `Data-test` dataset.

The analysis includes:

- 3-class evaluation for `Human`, `AI`, and `AI-assisted`
- Human-vs-AI evaluation
- 3-class confusion matrices
- 2-class confusion matrices
- overall model comparison metrics
- a fixed 20-sample comparison shared across all models
- per-model class probabilities for the same 20 samples
- number of models that misclassified each selected sample
- pairwise model agreement across the complete OOD test set

The following models are evaluated:

- DistilBERT
- GeLECTRA
- mmBERT-base
- mmBERT-small
- XLM-R

All confusion matrices and comparison tables are saved automatically in the project `results` directory.

```python
from pathlib import Path
import gc

import numpy as np
import pandas as pd
import torch
import matplotlib.pyplot as plt

from tqdm.auto import tqdm

from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    f1_score,
)

from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification,
    PreTrainedTokenizerFast,
)

PROJECT_DIR = Path(
    "german-ai-text-detector"
)

DATA_TEST_DIR = PROJECT_DIR / "Data-test"

RESULTS_DIR = PROJECT_DIR / "results"
RESULTS_3CLASS_DIR = RESULTS_DIR / "3-class-matrices"
RESULTS_2CLASS_DIR = RESULTS_DIR / "2-class-matrices"

RESULTS_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

RESULTS_3CLASS_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

RESULTS_2CLASS_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

BATCH_SIZE = 8
MAX_LENGTH = 512
RANDOM_SEED = 17

LABEL_NAMES = {
    0: "Human",
    1: "AI",
    2: "AI-assisted",
}

CLASS_NAMES = [
    "Human",
    "AI",
    "AI-assisted",
]

BINARY_NAMES = [
    "Human",
    "AI",
]

device = torch.device(
    "cuda"
    if torch.cuda.is_available()
    else "cpu"
)

MODEL_DIRS = {
    "DistilBERT": (
        PROJECT_DIR
        / "distilbert-german"
    ),

    "GeLECTRA": (
        PROJECT_DIR
        / "gelectra-large"
        / "model"
    ),

    "mmBERT-base": (
        PROJECT_DIR
        / "mmbert-base"
        / "mmbert-base-3class"
    ),

    "mmBERT-small": (
        PROJECT_DIR
        / "mmbert-small"
        / "model"
    ),

    "XLM-R": (
        PROJECT_DIR
        / "xlm-roberta-base"
        / "model"
    ),
}

for model_name, model_dir in MODEL_DIRS.items():

    if not model_dir.exists():

        raise FileNotFoundError(
            f"{model_name} not found:\n"
            f"{model_dir}"
        )

files_and_labels = []

for path in sorted(
    DATA_TEST_DIR.glob("*.txt")
):

    files_and_labels.append(
        (path, 0)
    )

for path in sorted(
    (DATA_TEST_DIR / "moderate").glob("*.txt")
):

    files_and_labels.append(
        (path, 1)
    )

for path in sorted(
    (DATA_TEST_DIR / "light").glob("*.txt")
):

    files_and_labels.append(
        (path, 2)
    )

all_paths = np.asarray(
    [
        path
        for path, _ in files_and_labels
    ],
    dtype=object,
)

true_labels = np.asarray(
    [
        label
        for _, label in files_and_labels
    ],
    dtype=np.int64,
)

all_texts = np.asarray(
    [
        path.read_text(
            encoding="utf-8",
            errors="ignore",
        )
        for path in all_paths
    ],
    dtype=object,
)

rng = np.random.default_rng(
    RANDOM_SEED
)

sample_counts = {
    0: 7,
    1: 7,
    2: 6,
}

comparison_indices = []

for label, count in sample_counts.items():

    label_indices = np.where(
        true_labels == label
    )[0]

    selected = rng.choice(
        label_indices,
        size=count,
        replace=False,
    )

    comparison_indices.extend(
        selected.tolist()
    )

comparison_indices = np.asarray(
    comparison_indices
)

prediction_table = pd.DataFrame(
    {
        "File": [
            all_paths[i].name
            for i in comparison_indices
        ],

        "Actual": [
            LABEL_NAMES[
                int(true_labels[i])
            ]
            for i in comparison_indices
        ],
    }
)

probability_table = pd.DataFrame(
    {
        "File": [
            all_paths[i].name
            for i in comparison_indices
        ],

        "Actual": [
            LABEL_NAMES[
                int(true_labels[i])
            ]
            for i in comparison_indices
        ],
    }
)

def load_tokenizer(
    model_dir,
):

    try:

        tokenizer = (
            AutoTokenizer
            .from_pretrained(
                model_dir,
                use_fast=True,
            )
        )

        return tokenizer

    except Exception as auto_error:

        tokenizer_json = (
            model_dir
            / "tokenizer.json"
        )

        if not tokenizer_json.exists():

            raise auto_error

        tokenizer = (
            PreTrainedTokenizerFast(
                tokenizer_file=str(
                    tokenizer_json
                )
            )
        )

        backend = (
            tokenizer.backend_tokenizer
        )

        token_0 = backend.id_to_token(0)
        token_1 = backend.id_to_token(1)
        token_2 = backend.id_to_token(2)

        if token_0 is not None:
            tokenizer.pad_token = token_0

        if token_1 is not None:
            tokenizer.eos_token = token_1

        if token_2 is not None:
            tokenizer.bos_token = token_2

        return tokenizer


def predict_model(
    model,
    tokenizer,
    texts,
):

    predictions = []
    probabilities = []

    for start in tqdm(
        range(
            0,
            len(texts),
            BATCH_SIZE,
        ),
        leave=False,
    ):

        batch_texts = (
            texts[
                start:start + BATCH_SIZE
            ]
            .tolist()
        )

        inputs = tokenizer(
            batch_texts,
            padding=True,
            truncation=True,
            max_length=MAX_LENGTH,
            return_tensors="pt",
        ).to(device)

        with torch.no_grad():

            outputs = model(
                **inputs
            )

            probs = torch.softmax(
                outputs.logits,
                dim=-1,
            )

        predictions.extend(
            probs
            .argmax(dim=-1)
            .cpu()
            .numpy()
        )

        probabilities.extend(
            probs
            .cpu()
            .numpy()
        )

    return (
        np.asarray(
            predictions,
            dtype=np.int64,
        ),

        np.asarray(
            probabilities,
        ),
    )


def plot_and_save_confusion_matrix(
    matrix,
    labels,
    title,
    figsize,
    output_file,
):

    fig, ax = plt.subplots(
        figsize=figsize
    )

    im = ax.imshow(
        matrix,
        cmap="Blues",
        vmin=0,
        vmax=matrix.max(),
    )

    positions = list(
        range(len(labels))
    )

    ax.set_xticks(
        positions,
        labels=labels,
    )

    ax.set_yticks(
        positions,
        labels=labels,
    )

    ax.set_xlabel(
        "Predicted label",
        fontsize=12,
    )

    ax.set_ylabel(
        "True label",
        fontsize=12,
    )

    ax.set_title(
        title,
        fontsize=14,
        fontweight="bold",
        pad=15,
    )

    threshold = (
        matrix.max() / 2
        if matrix.max() > 0
        else 0
    )

    for row in range(
        matrix.shape[0]
    ):

        for column in range(
            matrix.shape[1]
        ):

            value = matrix[
                row,
                column,
            ]

            ax.text(
                column,
                row,
                f"{value:,}",
                ha="center",
                va="center",
                fontsize=13,
                fontweight="bold",
                color=(
                    "white"
                    if value > threshold
                    else "black"
                ),
            )

    fig.colorbar(
        im,
        ax=ax,
        fraction=0.046,
        pad=0.04,
    )

    plt.tight_layout()

    fig.savefig(
        output_file,
        format="svg",
        bbox_inches="tight",
    )

    plt.show()

    plt.close(fig)


summary_rows = []

all_predictions = {}
all_probabilities = {}

for model_name, model_dir in MODEL_DIRS.items():

    tokenizer = load_tokenizer(
        model_dir
    )

    model = (
        AutoModelForSequenceClassification
        .from_pretrained(
            model_dir
        )
    )

    model.to(device)
    model.eval()

    predictions, probabilities = (
        predict_model(
            model,
            tokenizer,
            all_texts,
        )
    )

    all_predictions[
        model_name
    ] = predictions.copy()

    all_probabilities[
        model_name
    ] = probabilities.copy()

    accuracy_3class = (
        accuracy_score(
            true_labels,
            predictions,
        )
    )

    macro_f1 = f1_score(
        true_labels,
        predictions,
        labels=[0, 1, 2],
        average="macro",
        zero_division=0,
    )

    error_rates = {}

    for label in [0, 1, 2]:

        mask = (
            true_labels == label
        )

        error_rates[label] = (
            np.mean(
                predictions[mask]
                != label
            )
        )

    cm_3class = confusion_matrix(
        true_labels,
        predictions,
        labels=[0, 1, 2],
    )

    safe_name = (
        model_name
        .lower()
        .replace(" ", "-")
        .replace("+", "")
    )

    output_3class = (
        RESULTS_3CLASS_DIR
        / f"{safe_name}-3class.svg"
    )

    plot_and_save_confusion_matrix(
        cm_3class,
        CLASS_NAMES,
        (
            f"{model_name} — Data-test\n"
            "3-Class Confusion Matrix"
        ),
        figsize=(7, 6),
        output_file=output_3class,
    )

    binary_mask = np.isin(
        true_labels,
        [0, 1],
    )

    y_true_binary = (
        true_labels[
            binary_mask
        ]
    )

    y_pred_binary = np.where(
        predictions[
            binary_mask
        ] == 0,
        0,
        1,
    )

    accuracy_binary = (
        accuracy_score(
            y_true_binary,
            y_pred_binary,
        )
    )

    cm_binary = confusion_matrix(
        y_true_binary,
        y_pred_binary,
        labels=[0, 1],
    )

    output_2class = (
        RESULTS_2CLASS_DIR
        / f"{safe_name}-human-vs-ai.svg"
    )

    plot_and_save_confusion_matrix(
        cm_binary,
        BINARY_NAMES,
        (
            f"{model_name} — Data-test\n"
            "Human vs AI Confusion Matrix"
        ),
        figsize=(6, 5),
        output_file=output_2class,
    )

    prediction_table[
        model_name
    ] = [
        LABEL_NAMES[
            int(predictions[i])
        ]
        for i in comparison_indices
    ]

    selected_probabilities = (
        probabilities[
            comparison_indices
        ]
    )

    probability_table[
        f"{model_name} Human %"
    ] = (
        selected_probabilities[:, 0]
        * 100
    ).round(2)

    probability_table[
        f"{model_name} AI %"
    ] = (
        selected_probabilities[:, 1]
        * 100
    ).round(2)

    probability_table[
        f"{model_name} AI-assisted %"
    ] = (
        selected_probabilities[:, 2]
        * 100
    ).round(2)

    summary_rows.append(
        {
            "Model":
                model_name,

            "3-Class Accuracy":
                accuracy_3class,

            "Macro F1":
                macro_f1,

            "Human Error":
                error_rates[0],

            "AI Error":
                error_rates[1],

            "AI-assisted Error":
                error_rates[2],

            "Human vs AI Accuracy":
                accuracy_binary,
        }
    )

    del model
    del tokenizer

    gc.collect()

    if torch.cuda.is_available():

        torch.cuda.empty_cache()


model_columns = list(
    MODEL_DIRS.keys()
)

prediction_table[
    "Models Wrong"
] = (
    prediction_table[
        model_columns
    ]
    .ne(
        prediction_table["Actual"],
        axis=0,
    )
    .sum(
        axis=1
    )
)

summary_df = pd.DataFrame(
    summary_rows
)

summary_df = (
    summary_df
    .sort_values(
        "3-Class Accuracy",
        ascending=False,
    )
    .reset_index(
        drop=True
    )
)

agreement_matrix = pd.DataFrame(
    index=model_columns,
    columns=model_columns,
    dtype=float,
)

for model_a in model_columns:

    for model_b in model_columns:

        agreement_matrix.loc[
            model_a,
            model_b,
        ] = (
            np.mean(
                all_predictions[
                    model_a
                ]
                ==
                all_predictions[
                    model_b
                ]
            )
            * 100
        )

agreement_matrix = (
    agreement_matrix
    .round(2)
)

display(
    summary_df.style.format(
        {
            "3-Class Accuracy":
                "{:.2%}",

            "Macro F1":
                "{:.4f}",

            "Human Error":
                "{:.2%}",

            "AI Error":
                "{:.2%}",

            "AI-assisted Error":
                "{:.2%}",

            "Human vs AI Accuracy":
                "{:.2%}",
        }
    )
)

display(
    prediction_table
)

display(
    probability_table
)

display(
    agreement_matrix.style.format(
        "{:.2f}%"
    )
)

summary_df.to_csv(
    RESULTS_DIR
    / "model-comparison.csv",
    index=False,
)

prediction_table.to_csv(
    RESULTS_DIR
    / "20-sample-prediction-comparison.csv",
    index=False,
)

probability_table.to_csv(
    RESULTS_DIR
    / "20-sample-model-probabilities.csv",
    index=False,
)

agreement_matrix.to_csv(
    RESULTS_DIR
    / "model-agreement-matrix.csv",
    index=True,
)
```
